The raw data included common industry and sector classifications. The `IndustryId` field maps to the same field in the `us_companies` table.

```python
query = """
WITH cte AS (
SELECT *,
ROW_NUMBER() OVER (PARTITION BY Sector ORDER BY IndustryId) AS rn
FROM industries
)
SELECT Sector, Industry, IndustryId
FROM cte
WHERE rn in (1, 2, 3)
LIMIT 5;
"""

run_sql(query)
```

The company profile table (`us_companies`) had 37 rows where both the Ticker and Company name were blank (''). 

```python
query = """
SELECT * FROM us_companies;

SELECT COUNT(*)
FROM us_companies
WHERE (
        Ticker = '' OR Ticker IS NULL
    )
    AND (
        CompanyName = '' OR CompanyName IS NULL
    )
    AND (
        CIK = '' OR CIK IS NULL
    );

SELECT "CIK" AS Field, COUNT(*) AS "Missing Values"
FROM us_companies
WHERE CIK = '' OR CIK IS NULL

UNION

SELECT "Ticker" AS Field, COUNT(*) AS "Missing Values"
FROM us_companies
WHERE Ticker = '' OR Ticker IS NULL

UNION

SELECT "CompanyName" AS Field, COUNT(*) AS "Missing Values"
FROM us_companies
WHERE CompanyName = '' OR CompanyName IS NULL

UNION

SELECT "All" AS Field, COUNT(*) AS "Missing Values"
FROM us_companies
WHERE (
        CompanyName = '' OR CompanyName IS NULL
    )
    AND (
        Ticker = '' OR Ticker IS NULL
    )
    AND (
        CIK = '' OR CIK IS NULL
    );

    """
    
summary_missing_values = run_sql(query)
summary_missing_values.style.hide(axis="index")
```

The data includes several financial statement tables. The range of dates for all balance sheet, income statement, and cash flow tables were confirmed below.

```python

query = """
SELECT 
    "us_balance_quarterly" AS table_name,
    MIN(ReportDate) as min_date,
    MAX(ReportDate) as max_date,
    TIMESTAMPDIFF(DAY, MIN(ReportDate), MAX(ReportDate)) / 365.25 AS years
FROM us_balance_quarterly
UNION ALL
SELECT 
    "us_balance_annual" AS table_name,
    MIN(ReportDate) as min_date,
    MAX(ReportDate) as max_date,
    TIMESTAMPDIFF(DAY, MIN(ReportDate), MAX(ReportDate)) / 365.25 AS years
FROM us_balance_annual
UNION ALL
SELECT 
    "us_balance_ttm" AS table_name,
    MIN(ReportDate) as min_date,
    MAX(ReportDate) as max_date,
    TIMESTAMPDIFF(DAY, MIN(ReportDate), MAX(ReportDate)) / 365.25 AS years
FROM us_balance_ttm
UNION ALL
SELECT 
    "us_income_quarterly" AS table_name,
    MIN(ReportDate) as min_date,
    MAX(ReportDate) as max_date,
    TIMESTAMPDIFF(DAY, MIN(ReportDate), MAX(ReportDate)) / 365.25 AS years
FROM us_income_quarterly
UNION ALL
SELECT 
    "us_income_annual" AS table_name,
    MIN(ReportDate) as min_date,
    MAX(ReportDate) as max_date,
    TIMESTAMPDIFF(DAY, MIN(ReportDate), MAX(ReportDate)) / 365.25 AS years
FROM us_income_annual
UNION ALL
SELECT 
    "us_income_ttm" AS table_name,
    MIN(ReportDate) as min_date,
    MAX(ReportDate) as max_date,
    TIMESTAMPDIFF(DAY, MIN(ReportDate), MAX(ReportDate)) / 365.25 AS years
FROM us_income_ttm
UNION ALL
SELECT 
    "us_cashflow_quarterly" AS table_name,
    MIN(ReportDate) as min_date,
    MAX(ReportDate) as max_date,
    TIMESTAMPDIFF(DAY, MIN(ReportDate), MAX(ReportDate)) / 365.25 AS years
FROM us_cashflow_quarterly
UNION ALL
SELECT 
    "us_cashflow_annual" AS table_name,
    MIN(ReportDate) as min_date,
    MAX(ReportDate) as max_date,
    TIMESTAMPDIFF(DAY, MIN(ReportDate), MAX(ReportDate)) / 365.25 AS years
FROM us_cashflow_annual
UNION ALL
SELECT 
    "us_cashflow_ttm" AS table_name,        
    MIN(ReportDate) as min_date,
    MAX(ReportDate) as max_date,
    TIMESTAMPDIFF(DAY, MIN(ReportDate), MAX(ReportDate)) / 365.25 AS years
FROM us_cashflow_ttm;
"""

run_sql(query)
```

```python
# Hide
query = """
DROP TABLE IF EXISTS us_shareprices_daily_filtered;

CREATE TABLE us_shareprices_daily_filtered AS

WITH
    filtered_data AS (
        SELECT DISTINCT
            Ticker
        FROM us_shareprices_daily
        WHERE
            Close > 100000
            OR AdjClose > 100000
    )
SELECT *
FROM us_shareprices_daily
WHERE
    Ticker NOT IN (SELECT Ticker FROM filtered_data)
"""

run_sql(query)
```

# Ratios & Metrics

## Dates Lookup Table

Dates are an important part of any analysis that uses fundamentals data. Market price dates don't always align with fundamentals dates. Moreover, there are many kinds of dates to consider. Any serious analysis would try to avoid lookback bias where a decision based on information not yet known. To avoid this bias, the stock price date following the fundamentals publish date was collected. 

For backtesting decisions, it is often easier to use a regular month-end price date where the month-end date is the end of the month following the publish date. The month-end date following the publish date was calculated as the last day of the publish-date month or, if the publish date is within 3 days of the month end, the last day of the following month. 


```python
run_sql_file(SQL_DIR / "derived" / "create table - dates_lookup.sql")
dates_lookup = run_sql("SELECT * FROM dates_lookup")
display(dates_lookup.tail(5))
```

Profitability ratios measure a company’s ability to convert revenue and assets into profit. They reveal the strength of the firm’s core business model and its capacity to generate returns for shareholders. Metrics like gross margin, operating margin, net margin, ROA, and ROE help distinguish companies that merely grow revenue from those that grow value. These ratios also highlight competitive advantages, pricing power, cost discipline, and long‑term sustainability. In downstream analysis, profitability ratios often serve as leading indicators of strategic success and capital allocation effectiveness.

## Efficiency (Activity) Ratios

Efficiency ratios assess how effectively a company uses its assets and manages its operations. They highlight the speed and productivity with which resources are converted into revenue. Ratios such as asset turnover, inventory turnover, receivables turnover, and payables turnover illuminate operational discipline, supply‑chain management, and customer‑collection performance. These metrics often uncover bottlenecks or strengths that profitability ratios alone cannot reveal. Efficiency ratios are especially valuable for operational benchmarking and optimization.

```python
run_sql_file(SQL_DIR / "derived" / "create table - metrics_activity.sql")
metrics_activity = run_sql("SELECT * FROM metrics_activity")
metrics_activity.tail(5)
```

## Valuation Ratios

Valuation ratios connect fundamentals to market expectations. They measure how investors price a company relative to its earnings, cash flow, book value, or sales. Ratios like P/E, P/B, P/S, and EV/EBITDA help determine whether a stock appears undervalued, fairly valued, or overvalued. These metrics reflect sentiment, growth expectations, and perceived risk. In downstream analysis, valuation ratios bridge the gap between internal performance and external market behavior, enabling more informed investment decisions.

```python
run_sql_file(SQL_DIR / "derived" / "create table - metrics_value.sql")
metrics_value = run_sql("SELECT * FROM metrics_value")
metrics_value.tail(5)
```

## Liquidity Ratios

Liquidity ratios evaluate a company’s ability to meet short‑term obligations using its most liquid assets. They answer a fundamental question: Can the company pay its bills without stress? Ratios such as the current ratio, quick ratio, and cash ratio help identify operational resilience and working‑capital efficiency. Strong liquidity supports stability during downturns, while weak liquidity can signal cash‑flow strain or reliance on external financing. These ratios are essential for risk assessment, especially in volatile industries or early‑stage companies.

```python
run_sql_file(SQL_DIR / "derived" / "create table - metrics_liquidity.sql")
liquidity = run_sql("SELECT * FROM metrics_liquidity")
liquidity.tail(5)
```

## Solvency Ratios

Solvency ratios measure long‑term financial stability and the degree to which a company relies on debt to finance operations. Metrics like debt‑to‑equity, interest coverage, and debt‑to‑assets reveal leverage levels and the firm’s ability to service debt over time. High leverage can amplify returns but also magnify risk. Solvency ratios help analysts understand capital structure, creditworthiness, and exposure to macroeconomic conditions. They are critical for evaluating long‑term viability and strategic flexibility.

```python
run_sql_file(SQL_DIR / "derived" / "create table - metrics_solvency.sql")
metrics_solvency = run_sql("SELECT * FROM metrics_solvency")
metrics_solvency.tail(5)
```

## Cash Flow Quality Ratios

Cash flow quality reflects how reliably a company turns its accounting earnings into actual cash, offering a deeper look into the sustainability of its operations beyond reported profit figures; metrics such as Earnings Quality, FCF Margin, and FCF Conversion reveal whether earnings are supported by cash generation or distorted by accruals, capital intensity, or working‑capital swings, and across the companies shown, ZUO demonstrates weak cash flow quality with negative earnings support and limited free‑cash generation, ZYME shows unusually strong earnings quality but poor FCF margin - often a sign of heavy investment or non‑recurring cash movements - and ZYNE consistently converts earnings into cash effectively despite missing FCF margin data, suggesting disciplined cash management; overall, cash flow quality acts as a critical lens for distinguishing durable performance from accounting noise, identifying firms capable of self‑funding growth, and assessing resilience when paired with profitability, efficiency, liquidity, and solvency ratios.

```python
run_sql_file(SQL_DIR / "derived" / "create table - metrics_earnings_quality.sql")
metrics_earnings_quality = run_sql("SELECT * FROM metrics_earnings_quality")
metrics_earnings_quality.tail(5)
```

# Further Ideas
The data output from this project are informative and set the stage for further analysis. However, there are additional improvements that could be made to improve the efficiency of the data acquisition, processing, and output quality. They include:
1. Fetching the raw data from SimFin via its native API. This could involve automated download, unzipping, and folder cleanup.
2. Handling outlier values. For example, the stock prices contained extreme, probably incorrect data values. These could be deleted to ensure they do not affect subsequent calculations.
3. Handling financial report restatements. Occasionally, financial report restatements occur and the date of the restatement should become the effective date of the reported financial figures.
